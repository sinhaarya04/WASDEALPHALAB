# Step 2: Build Base Tables

This notebook builds clean base tables for each commodity (corn, soy, wheat) from raw futures CSV files.

**Rules:**
- Do NOT add features
- Do NOT do ML
- Do NOT join WASDE or macro data yet
- This step only builds clean merged base tables


```python
import pandas as pd
import numpy as np
```

```python
BASE_PATH = "/Users/aryansinha/Desktop/WASDA"

CORN_PATH  = f"{BASE_PATH}/raw/futures/corn"
SOY_PATH   = f"{BASE_PATH}/raw/futures/soybean"
WHEAT_PATH = f"{BASE_PATH}/raw/futures/wheat"
```

## TASK 4 — BUILD CORN BASE TABLE


### 4.1 Load raw corn CSVs

Note: CSV files have 6 header rows, so we skip them when reading.


```python
# Load corn CSVs (skip first 6 rows which contain metadata)
c1_bid = pd.read_csv(f"{CORN_PATH}/C1_BID(Worksheet).csv", skiprows=6)
c1_oi  = pd.read_csv(f"{CORN_PATH}/C1_OI(Worksheet).csv", skiprows=6)
c1_vol = pd.read_csv(f"{CORN_PATH}/C1_VOL(Worksheet).csv", skiprows=6)

c2_bid = pd.read_csv(f"{CORN_PATH}/C2_BIDPRICE(Worksheet).csv", skiprows=6)
c2_oi  = pd.read_csv(f"{CORN_PATH}/C2_OI(Worksheet).csv", skiprows=6)
c2_vol = pd.read_csv(f"{CORN_PATH}/C2_VOL(Worksheet).csv", skiprows=6)
```

### 4.2 Inspect each dataframe


```python
# Inspect c1_bid
print("=== C1_BID ===")
print(c1_bid.head())
print("\n")
print(c1_bid.info())
print("\n" + "="*50 + "\n")
```

```python
# Inspect c1_oi
print("=== C1_OI ===")
print(c1_oi.head())
print("\n")
print(c1_oi.info())
print("\n" + "="*50 + "\n")
```

```python
# Inspect c1_vol
print("=== C1_VOL ===")
print(c1_vol.head())
print("\n")
print(c1_vol.info())
print("\n" + "="*50 + "\n")
```

```python
# Inspect c2_bid
print("=== C2_BID ===")
print(c2_bid.head())
print("\n")
print(c2_bid.info())
print("\n" + "="*50 + "\n")
```

```python
# Inspect c2_oi
print("=== C2_OI ===")
print(c2_oi.head())
print("\n")
print(c2_oi.info())
print("\n" + "="*50 + "\n")
```

```python
# Inspect c2_vol
print("=== C2_VOL ===")
print(c2_vol.head())
print("\n")
print(c2_vol.info())
print("\n" + "="*50 + "\n")
```

### 4.3 Standardize columns

Rename columns so every dataframe has: `date | value_column`


```python
# Standardize c1_bid: Date -> date, PX_BID -> c1_bid
c1_bid = c1_bid.rename(columns={
    "Date": "date",
    "PX_BID": "c1_bid"
})
# Keep only date and value column
c1_bid = c1_bid[["date", "c1_bid"]]

print("C1_BID columns:", c1_bid.columns.tolist())
print(c1_bid.head())
```

```python
# Standardize c1_oi: Date -> date, OPEN_INT -> c1_oi
c1_oi = c1_oi.rename(columns={
    "Date": "date",
    "OPEN_INT": "c1_oi"
})
c1_oi = c1_oi[["date", "c1_oi"]]

print("C1_OI columns:", c1_oi.columns.tolist())
print(c1_oi.head())
```

```python
# Standardize c1_vol: Date -> date, PX_VOLUME -> c1_vol
c1_vol = c1_vol.rename(columns={
    "Date": "date",
    "PX_VOLUME": "c1_vol"
})
c1_vol = c1_vol[["date", "c1_vol"]]

print("C1_VOL columns:", c1_vol.columns.tolist())
print(c1_vol.head())
```

```python
# Standardize c2_bid: Date -> date, PX_BID -> c2_bid
c2_bid = c2_bid.rename(columns={
    "Date": "date",
    "PX_BID": "c2_bid"
})
c2_bid = c2_bid[["date", "c2_bid"]]

print("C2_BID columns:", c2_bid.columns.tolist())
print(c2_bid.head())
```

```python
# Standardize c2_oi: Date -> date, OPEN_INT -> c2_oi
c2_oi = c2_oi.rename(columns={
    "Date": "date",
    "OPEN_INT": "c2_oi"
})
c2_oi = c2_oi[["date", "c2_oi"]]

print("C2_OI columns:", c2_oi.columns.tolist())
print(c2_oi.head())
```

```python
# Standardize c2_vol: Date -> date, PX_VOLUME -> c2_vol
c2_vol = c2_vol.rename(columns={
    "Date": "date",
    "PX_VOLUME": "c2_vol"
})
c2_vol = c2_vol[["date", "c2_vol"]]

print("C2_VOL columns:", c2_vol.columns.tolist())
print(c2_vol.head())
```

```python
# Convert all dates to datetime
for df in [c1_bid, c1_oi, c1_vol, c2_bid, c2_oi, c2_vol]:
    df["date"] = pd.to_datetime(df["date"])

print("Dates converted to datetime")
print("Sample dates:")
print(c1_bid["date"].head())
```

### 4.4 Merge into ONE corn table

Merge step-by-step on `date`


```python
# Merge all corn dataframes on date
corn = c1_bid.merge(c1_oi,  on="date", how="inner")
corn = corn.merge(c1_vol, on="date", how="inner")
corn = corn.merge(c2_bid, on="date", how="inner")
corn = corn.merge(c2_oi,  on="date", how="inner")
corn = corn.merge(c2_vol, on="date", how="inner")

# Sort by date and reset index
corn = corn.sort_values("date").reset_index(drop=True)

print("Corn base table shape:", corn.shape)
print("Columns:", corn.columns.tolist())
```

### 4.5 Sanity checks


```python
# Display head and tail
print("=== CORN BASE TABLE - HEAD ===")
print(corn.head())
print("\n=== CORN BASE TABLE - TAIL ===")
print(corn.tail())
print("\n=== MISSING VALUES ===")
print(corn.isna().sum())
print("\n=== DATE RANGE ===")
print(f"Start: {corn['date'].min()}")
print(f"End: {corn['date'].max()}")
print(f"Total rows: {len(corn)}")
```

### 4.6 Save output

Expected columns: `date | c1_bid | c1_oi | c1_vol | c2_bid | c2_oi | c2_vol`


```python
# Save corn base table
corn.to_csv(
    f"{BASE_PATH}/processed/base_tables/corn_base.csv",
    index=False
)

print(f"✅ Corn base table saved to: {BASE_PATH}/processed/base_tables/corn_base.csv")
print(f"Final shape: {corn.shape}")
print(f"Final columns: {corn.columns.tolist()}")
```

## TASK 5 — BUILD SOYBEAN BASE TABLE


### 5.1 Load raw soybean CSVs

Note: CSV files have 6 header rows, so we skip them when reading.


```python
# Load soybean CSVs (skip first 6 rows which contain metadata)
s1_bid = pd.read_csv(f"{SOY_PATH}/S1_BID(Worksheet).csv", skiprows=6)
s1_oi  = pd.read_csv(f"{SOY_PATH}/S1_OI(Worksheet).csv", skiprows=6)
s1_vol = pd.read_csv(f"{SOY_PATH}/s1_vol(Worksheet).csv", skiprows=6)  # Note: lowercase filename

s2_bid = pd.read_csv(f"{SOY_PATH}/S2_BID(Worksheet).csv", skiprows=6)
s2_oi  = pd.read_csv(f"{SOY_PATH}/S2_OI(Worksheet).csv", skiprows=6)
s2_vol = pd.read_csv(f"{SOY_PATH}/S2_VOL(Worksheet).csv", skiprows=6)
```

### 5.2 Standardize columns

Rename columns so every dataframe has: `date | value_column`


```python
# Standardize s1_bid: Date -> date, PX_BID -> s1_bid
s1_bid = s1_bid.rename(columns={
    "Date": "date",
    "PX_BID": "s1_bid"
})
s1_bid = s1_bid[["date", "s1_bid"]]

# Standardize s1_oi: Date -> date, OPEN_INT -> s1_oi
s1_oi = s1_oi.rename(columns={
    "Date": "date",
    "OPEN_INT": "s1_oi"
})
s1_oi = s1_oi[["date", "s1_oi"]]

# Standardize s1_vol: Date -> date, PX_VOLUME -> s1_vol
s1_vol = s1_vol.rename(columns={
    "Date": "date",
    "PX_VOLUME": "s1_vol"
})
s1_vol = s1_vol[["date", "s1_vol"]]

# Standardize s2_bid: Date -> date, PX_BID -> s2_bid
s2_bid = s2_bid.rename(columns={
    "Date": "date",
    "PX_BID": "s2_bid"
})
s2_bid = s2_bid[["date", "s2_bid"]]

# Standardize s2_oi: Date -> date, OPEN_INT -> s2_oi
s2_oi = s2_oi.rename(columns={
    "Date": "date",
    "OPEN_INT": "s2_oi"
})
s2_oi = s2_oi[["date", "s2_oi"]]

# Standardize s2_vol: Date -> date, PX_VOLUME -> s2_vol
s2_vol = s2_vol.rename(columns={
    "Date": "date",
    "PX_VOLUME": "s2_vol"
})
s2_vol = s2_vol[["date", "s2_vol"]]

# Convert all dates to datetime
for df in [s1_bid, s1_oi, s1_vol, s2_bid, s2_oi, s2_vol]:
    df["date"] = pd.to_datetime(df["date"])

print("Soybean columns standardized")
print("Sample columns:")
print("s1_bid:", s1_bid.columns.tolist())
print("s1_oi:", s1_oi.columns.tolist())
print("s1_vol:", s1_vol.columns.tolist())
```

### 5.3 Merge into ONE soybean table


```python
# Merge all soybean dataframes on date
soy = s1_bid.merge(s1_oi,  on="date", how="inner")
soy = soy.merge(s1_vol, on="date", how="inner")
soy = soy.merge(s2_bid, on="date", how="inner")
soy = soy.merge(s2_oi,  on="date", how="inner")
soy = soy.merge(s2_vol, on="date", how="inner")

# Sort by date and reset index
soy = soy.sort_values("date").reset_index(drop=True)

print("Soybean base table shape:", soy.shape)
print("Columns:", soy.columns.tolist())
```

### 5.4 Sanity checks


```python
# Display head and tail
print("=== SOYBEAN BASE TABLE - HEAD ===")
print(soy.head())
print("\n=== SOYBEAN BASE TABLE - TAIL ===")
print(soy.tail())
print("\n=== MISSING VALUES ===")
print(soy.isna().sum())
print("\n=== DATE RANGE ===")
print(f"Start: {soy['date'].min()}")
print(f"End: {soy['date'].max()}")
print(f"Total rows: {len(soy)}")
```

### 5.5 Save output

Expected columns: `date | s1_bid | s1_oi | s1_vol | s2_bid | s2_oi | s2_vol`


```python
# Save soybean base table
soy.to_csv(
    f"{BASE_PATH}/processed/base_tables/soybean_base.csv",
    index=False
)

print(f"✅ Soybean base table saved to: {BASE_PATH}/processed/base_tables/soybean_base.csv")
print(f"Final shape: {soy.shape}")
print(f"Final columns: {soy.columns.tolist()}")
```
